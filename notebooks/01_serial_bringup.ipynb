# 01 Serial bring-up

Purpose: find the AudioStar's USB-serial port and confirm the handbox answers.
The ACK byte (0x06) returns `A` (alt-az), `L` (land) or `P` (polar) per the Meade protocol doc.
Nothing here moves the telescope.

```python
import serial
from serial.tools import list_ports
```

```python
# List candidate ports; the old setup used a PL2303/FTDI adapter (/dev/tty.usbserial-*).
for p in list_ports.comports():
    print(p.device, '|', p.description)
```

```python
PORT = None  # set from the list above, e.g. '/dev/tty.usbserial-XXXX'
if PORT:
    with serial.Serial(PORT, 9600, timeout=1) as s:
        s.write(b'\x06')
        print('mount mode:', s.read(1))
        s.write(b':GVP#')
        print('product:', s.read_until(b'#'))
```
